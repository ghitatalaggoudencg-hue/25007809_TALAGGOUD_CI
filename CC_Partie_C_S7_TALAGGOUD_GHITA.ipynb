```python
# Sources réelles intégrées à scikit-learn : trois problèmes différents
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes, load_breast_cancer, load_wine
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, f1_score, roc_auc_score, silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
GRAINE = 42

jeu_regression = load_diabetes()
X_reg, y_reg = jeu_regression.data, jeu_regression.target
jeu_classification = load_breast_cancer()
X_cls, y_cls = jeu_classification.data, jeu_classification.target
jeu_non_supervise = load_wine()
X_clu = StandardScaler().fit_transform(jeu_non_supervise.data)  # Exploration sans cible

Xr_ent, Xr_test, yr_ent, yr_test = train_test_split(X_reg, y_reg, test_size=.25, random_state=GRAINE)
Xc_ent, Xc_test, yc_ent, yc_test = train_test_split(X_cls, y_cls, test_size=.25, stratify=y_cls, random_state=GRAINE)
print('Diabetes:', X_reg.shape, '| Breast Cancer:', X_cls.shape, '| Wine:', X_clu.shape)
print('Ces données sont réelles ; les étiquettes du vin ne sont pas employées dans le clustering.')
```

**Aides du travail (méthode) :** les réglages sont comparés par validation croisée sur l'entraînement uniquement, et chaque schéma est tracé par la fonction `schema`.

```python
# Aide : toutes les comparaisons d'hyperparamètres se font par validation croisée sur l'ENTRAÎNEMENT uniquement (le test n'est jamais utilisé pour régler)
import time
from sklearn.model_selection import cross_val_score, cross_val_predict, KFold, StratifiedKFold
cv_reg = KFold(5, shuffle=True, random_state=GRAINE)
cv_cls = StratifiedKFold(5, shuffle=True, random_state=GRAINE)
def cv_mae(modele):
    s = -cross_val_score(modele, Xr_ent, yr_ent, cv=cv_reg, scoring='neg_mean_absolute_error')
    return float(round(s.mean(), 2)), float(round(s.std(), 2))
def cv_f1(modele, scoring='f1'):
    s = cross_val_score(modele, Xc_ent, yc_ent, cv=cv_cls, scoring=scoring)
    return float(round(s.mean(), 3)), float(round(s.std(), 3))
print("Aides définies : cv_mae (MAE moyenne ± écart-type sur 5 plis) et cv_f1 (F1 moyen ± écart-type sur 5 plis).")
```

```python
import textwrap, matplotlib.patches as mpatches
def schema(titre, init, appr, sortie, choix):
    """Schéma original : ce qui change entre l'initialisation, l'apprentissage et la sortie."""
    fig, ax = plt.subplots(figsize=(9, 2.5)); ax.axis('off')
    for i, (nom, txt, c) in enumerate(zip(['1. Initialisation', '2. Apprentissage', '3. Sortie'], [init, appr, sortie], ['#e8def8', '#d6eaf8', '#d5f5e3'])):
        ax.add_patch(mpatches.FancyBboxPatch((i * 3.3, 0.9), 2.9, 1.6, boxstyle='round,pad=0.05', fc=c, ec='k'))
        ax.text(i * 3.3 + 1.45, 2.25, nom, ha='center', fontweight='bold', fontsize=9)
        ax.text(i * 3.3 + 1.45, 1.5, textwrap.fill(txt, 28), ha='center', va='center', fontsize=8)
        if i < 2: ax.annotate('', xy=((i + 1) * 3.3 - 0.1, 1.7), xytext=(i * 3.3 + 3.0, 1.7), arrowprops=dict(arrowstyle='->', lw=1.5))
    ax.text(4.9, 0.4, 'Choix algorithmique illustré : ' + choix, ha='center', fontsize=9, style='italic')
    ax.set_xlim(-0.2, 9.9); ax.set_ylim(0, 2.9); ax.set_title(titre, fontsize=11); plt.show()
print('Fonction schema() définie.')
```

## 1. Régression linéaire

**Variante C — question spécifique :** Quel diagnostic du résidu indique une mauvaise spécification linéaire ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import LinearRegression
modele = LinearRegression().fit(Xr_ent, yr_ent)
predictions = modele.predict(Xr_test)
print('MAE régression linéaire :', round(mean_absolute_error(yr_test, predictions), 2))
print('R² :', round(r2_score(yr_test, predictions), 3))
```

**Présentation :** La régression linéaire est un algorithme supervisé qui reçoit des variables numériques X et prédit une valeur numérique y en supposant une relation linéaire, ŷ = β₀ + β₁x₁ + … + βₚxₚ (ŷ = valeur prédite, xⱼ = variables, β₀ = ordonnée à l'origine, βⱼ = coefficients), où les β sont les paramètres appris en minimisant la somme des carrés des erreurs, alors que fit_intercept est un hyperparamètre fixé avant l'apprentissage.

```python
schema('1. Régression linéaire', 'coefficients β inconnus (aucune droite)', "minimiser Σ(y − ŷ)² sur l'entraînement", 'droite ŷ = β₀ + β·x', 'la fonction objectif des moindres carrés')
```

**Schéma original :** Le schéma montre le passage de « coefficients β inconnus (aucune droite) » (initialisation) à « minimiser Σ(y − ŷ)² sur l'entraînement » (apprentissage) puis à « droite ŷ = β₀ + β·x » (sortie), ce qui illustre la fonction objectif des moindres carrés.

**Contextes et choix :** Elle sert à prévoir les ventes à l'export à partir du prix et du budget publicitaire (commerce international) ou le coût de transport à partir de la distance (logistique) sur un tableau de variables numériques, face à Ridge comme concurrent, et elle est mal adaptée à une relation courbe ou à des variables très corrélées, la MAE étant choisie comme métrique car elle s'exprime dans l'unité de la cible.

```python
from sklearn.linear_model import LinearRegression
for intercept in [True, False]:
    print('fit_intercept =', intercept, '| MAE CV (moy, écart-type) :', cv_mae(LinearRegression(fit_intercept=intercept)))
pred_cv = cross_val_predict(LinearRegression(), Xr_ent, yr_ent, cv=cv_reg)
residus = yr_ent - pred_cv
print('Corrélation résidu / prédiction² :', round(np.corrcoef(residus, pred_cv**2)[0, 1], 3))
bas, haut = pred_cv < np.median(pred_cv), pred_cv >= np.median(pred_cv)
print('Écart-type des résidus (prédictions basses / hautes) :', round(residus[bas].std(), 1), '/', round(residus[haut].std(), 1))
plt.scatter(pred_cv, residus, s=14); plt.axhline(0, color='k'); plt.xlabel('Prédiction (validation croisée)'); plt.ylabel('Résidu'); plt.title('Diabetes — résidus de la régression linéaire'); plt.show()
```

**Test de sensibilité et interprétation :** En passant de fit_intercept=True à False la MAE en validation croisée monte de 45.81 à 157.28, et le diagnostic d'une mauvaise spécification linéaire est un motif visible (courbe ou entonnoir) dans le graphique résidus/prédictions, ici faible (corrélation résidu/prédiction² = -0.024, écart-type des résidus 52.4 pour les prédictions basses contre 59.6 pour les hautes), mais une seule exécution sur un seul découpage ne prouve pas que la relation est vraiment linéaire.

## 2. Ridge

**Variante C — question spécifique :** Comment évoluent les coefficients quand alpha augmente ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), Ridge(alpha=12.0)).fit(Xr_ent, yr_ent)
print('MAE Ridge :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

**Présentation :** Ridge est une régression linéaire supervisée pénalisée qui minimise Σ(y − ŷ)² + α·Σβⱼ² (y = valeur réelle, ŷ = valeur prédite, βⱼ = coefficients appris, α = force de la pénalité, qui est l'hyperparamètre), en standardisant d'abord les variables puis en réduisant les coefficients vers zéro sans les annuler, avec l'hypothèse que des coefficients petits généralisent mieux.

```python
schema('2. Ridge', 'α fixé, coefficients β libres', 'minimiser erreurs² + α·Σβ²', 'coefficients rétrécis mais non nuls', 'la pénalité L2 ajoutée à la fonction objectif')
```

**Schéma original :** Le schéma montre le passage de « α fixé, coefficients β libres » (initialisation) à « minimiser erreurs² + α·Σβ² » (apprentissage) puis à « coefficients rétrécis mais non nuls » (sortie), ce qui illustre la pénalité L2 ajoutée à la fonction objectif.

**Contextes et choix :** Elle convient pour prédire un prix à partir de nombreux indicateurs de marché corrélés (marketing) ou un délai de livraison à partir de variables logistiques proches, sur un tableau numérique standardisé, face à Lasso comme concurrent, et elle est mal adaptée quand on veut éliminer des variables inutiles, la MAE étant la métrique.

```python
for alpha in [0.1, 12, 100, 1000]:
    m = make_pipeline(StandardScaler(), Ridge(alpha=alpha))
    print('alpha =', alpha, '| norme des coefficients :', round(np.linalg.norm(m.fit(Xr_ent, yr_ent)[-1].coef_), 2), '| MAE CV :', cv_mae(m))
```

**Test de sensibilité et interprétation :** Quand alpha passe de 0.1 à 1000 la norme des coefficients baisse de 69.02 à 15.16 mais la MAE CV monte de 45.8 à 56.01 (alpha=12 donne 45.91), donc plus alpha est grand plus les coefficients rétrécissent jusqu'au sous-apprentissage, et une seule exécution ne permet pas de dire que alpha=12 est optimal.

## 3. Lasso

**Variante C — question spécifique :** Combien de variables sont annulées pour trois valeurs d’alpha ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import Lasso
modele = make_pipeline(StandardScaler(), Lasso(alpha=2.0, max_iter=15000)).fit(Xr_ent, yr_ent)
print('Nombre de coefficients non nuls :', np.count_nonzero(modele[-1].coef_))
print('MAE Lasso :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

**Présentation :** Lasso est une régression linéaire supervisée pénalisée par la somme des valeurs absolues qui minimise (1/2n)·Σ(y − ŷ)² + α·Σ|βⱼ| (n = nombre d'observations, βⱼ = coefficients appris, α = pénalité, hyperparamètre) et qui met certains coefficients exactement à zéro, ce qui sélectionne les variables.

```python
schema('3. Lasso', 'α fixé, tous les β possibles', 'minimiser erreurs² + α·Σ|β|', 'certains β = 0 (variables éliminées)', 'la pénalité L1 qui sélectionne les variables')
```

**Schéma original :** Le schéma montre le passage de « α fixé, tous les β possibles » (initialisation) à « minimiser erreurs² + α·Σ|β| » (apprentissage) puis à « certains β = 0 (variables éliminées) » (sortie), ce qui illustre la pénalité L1 qui sélectionne les variables.

**Contextes et choix :** Il sert à repérer les indicateurs vraiment utiles parmi beaucoup de variables pour prévoir un risque de crédit (finance) ou la demande d'un produit (commerce), sur un tableau numérique standardisé, face à Ridge, et il est mal adapté quand toutes les variables corrélées sont utiles, la MAE étant la métrique.

```python
for alpha in [0.1, 2, 10]:
    m = make_pipeline(StandardScaler(), Lasso(alpha=alpha, max_iter=15000))
    print('alpha =', alpha, '| coefficients non nuls :', np.count_nonzero(m.fit(Xr_ent, yr_ent)[-1].coef_), 'sur', Xr_ent.shape[1], '| MAE CV :', cv_mae(m))
```

**Test de sensibilité et interprétation :** Avec alpha = 0.1, 2 puis 10 le modèle garde 9, 8 puis 4 coefficients sur 10 et la MAE CV passe de 45.75 à 46.54 puis 49.67, donc plus alpha est grand plus Lasso élimine de variables au prix d'une erreur plus forte, et un seul découpage ne dit pas quelles variables sont vraiment importantes.

## 4. Régression polynomiale

**Variante C — question spécifique :** Le degré 2 améliore-t-il le test ou seulement l’entraînement ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), PolynomialFeatures(degree=2, include_bias=False), Ridge(alpha=100.)).fit(Xr_ent, yr_ent)
print('R² polynomial :', round(r2_score(yr_test, modele.predict(Xr_test)), 3))
```

**Présentation :** La régression polynomiale est une régression supervisée qui ajoute des puissances et des produits des variables (x², x₁x₂…) avant un modèle linéaire, ŷ = β₀ + Σβⱼxⱼ + Σβⱼₖxⱼxₖ (ŷ = prédiction, x = variables, β = coefficients appris), le degré étant l'hyperparamètre qui règle la souplesse de la courbe.

```python
schema('4. Régression polynomiale', "variables d'origine x", 'créer x², x₁x₂… puis ajuster les β', 'courbe ou surface non linéaire', 'le degré (souplesse du modèle)')
```

**Schéma original :** Le schéma montre le passage de « variables d'origine x » (initialisation) à « créer x², x₁x₂… puis ajuster les β » (apprentissage) puis à « courbe ou surface non linéaire » (sortie), ce qui illustre le degré (souplesse du modèle).

**Contextes et choix :** Elle convient pour modéliser un effet qui sature, comme la réponse des ventes à un budget publicitaire (marketing) ou la consommation de carburant selon la vitesse (logistique), sur peu de variables numériques, face à l'arbre de régression, et elle est mal adaptée avec beaucoup de variables car le nombre de termes explose, le R² en validation servant de métrique.

```python
for degre in [1, 2, 3]:
    m = make_pipeline(StandardScaler(), PolynomialFeatures(degree=degre, include_bias=False), Ridge(alpha=100.))
    r2_ent = r2_score(yr_ent, m.fit(Xr_ent, yr_ent).predict(Xr_ent))
    r2_val = cross_val_score(m, Xr_ent, yr_ent, cv=cv_reg, scoring='r2').mean()
    print('degré =', degre, '| R² entraînement :', round(r2_ent, 3), '| R² validation croisée :', round(r2_val, 3))
```

**Test de sensibilité et interprétation :** En passant du degré 1 à 2 puis 3 le R² d'entraînement monte de 0.502 à 0.557 puis 0.662 alors que le R² de validation croisée baisse de 0.448 à 0.416 puis 0.222, donc le degré 2 n'améliore que l'entraînement (surajustement) et pas la validation, le test restant intact, et un seul résultat ne prouve pas qu'un autre alpha ne changerait rien.

## 5. Arbre de régression

**Variante C — question spécifique :** Quel effet de max_depth sur le biais et la variance ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.tree import DecisionTreeRegressor
modele = DecisionTreeRegressor(max_depth=3, min_samples_leaf=8, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE arbre :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

**Présentation :** L'arbre de régression est un algorithme supervisé qui coupe les données par des questions « variable < seuil ? » choisies pour minimiser l'erreur quadratique MSE = (1/n)·Σ(y − ȳ)² (n = observations du nœud, ȳ = moyenne des y du nœud) et qui prédit la moyenne de y dans chaque feuille, les variables et seuils étant appris et max_depth et min_samples_leaf étant des hyperparamètres.

```python
schema('5. Arbre de régression', 'un seul groupe (racine)', 'couper au seuil qui réduit le plus MSE', 'feuilles = moyennes de y', "le critère de séparation (réduction de l'erreur)")
```

**Schéma original :** Le schéma montre le passage de « un seul groupe (racine) » (initialisation) à « couper au seuil qui réduit le plus MSE » (apprentissage) puis à « feuilles = moyennes de y » (sortie), ce qui illustre le critère de séparation (réduction de l'erreur).

**Contextes et choix :** Il sert à estimer le montant d'un sinistre par des règles lisibles (assurance, finance) ou un coût de livraison par zone (logistique) sur un tableau mixte sans normalisation, face à la forêt aléatoire, et il est mal adapté quand on veut une prédiction stable car il change beaucoup avec les données, la MAE étant la métrique.

## 6. Forêt aléatoire — régression

**Variante C — question spécifique :** La forêt réduit-elle la variance par rapport à l’arbre ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.ensemble import RandomForestRegressor
modele = RandomForestRegressor(n_estimators=60, max_depth=6, n_jobs=-1, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE forêt :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

**Présentation :** La forêt aléatoire de régression est un algorithme supervisé qui construit plusieurs arbres sur des échantillons tirés au hasard (bootstrap) et prédit leur moyenne, ŷ = (1/B)·Σ arbre_b(x) (B = nombre d'arbres, arbre_b(x) = prédiction de l'arbre b), les coupures des arbres étant apprises et n_estimators et max_depth étant des hyperparamètres.

```python
schema('6. Forêt aléatoire — régression', 'B arbres vides + échantillons bootstrap', 'chaque arbre apprend sur son échantillon', 'moyenne des B prédictions', "l'agrégation (bagging) qui réduit la variance")
```

**Schéma original :** Le schéma montre le passage de « B arbres vides + échantillons bootstrap » (initialisation) à « chaque arbre apprend sur son échantillon » (apprentissage) puis à « moyenne des B prédictions » (sortie), ce qui illustre l'agrégation (bagging) qui réduit la variance.

**Contextes et choix :** Elle convient pour prévoir la demande d'un produit avec des relations non linéaires (commerce) ou le prix d'un bien (finance) sur un tableau de variables mixtes, face au gradient boosting, et elle est mal adaptée quand on doit expliquer chaque coefficient, la MAE étant la métrique.

```python
Xa, Xb, ya, yb = train_test_split(Xr_ent, yr_ent, test_size=.25, random_state=GRAINE)  # découpage interne à l'entraînement
def variance_predictions(fabrique, n_rep=15):
    rng = np.random.default_rng(GRAINE); preds = []
    for _ in range(n_rep):
        idx = rng.integers(0, len(Xa), len(Xa))
        preds.append(fabrique().fit(Xa[idx], ya[idx]).predict(Xb))
    return round(float(np.mean(np.var(preds, axis=0))), 1)
print('Arbre prof. 6 | MAE CV :', cv_mae(DecisionTreeRegressor(max_depth=6, min_samples_leaf=8, random_state=GRAINE)), '| variance des prédictions (rééchantillonnage) :', variance_predictions(lambda: DecisionTreeRegressor(max_depth=6, min_samples_leaf=8, random_state=GRAINE)))
for n in [10, 60, 200]:
    print('Forêt n_estimators =', n, '| MAE CV :', cv_mae(RandomForestRegressor(n_estimators=n, max_depth=6, n_jobs=-1, random_state=GRAINE)), '| variance des prédictions :', variance_predictions(lambda: RandomForestRegressor(n_estimators=n, max_depth=6, n_jobs=-1, random_state=GRAINE)))
```

**Test de sensibilité et interprétation :** Oui, la forêt réduit la variance : sur des rééchantillonnages la variance des prédictions passe de 1440.4 pour un arbre de profondeur 6 à 322.9 pour 60 arbres (459.4 pour 10 arbres et 298.7 pour 200) et la MAE CV passe de 57.63 à 50.22, mais ces valeurs viennent d'un seul découpage et ne garantissent pas le même gain sur d'autres données.

**Variante C — question spécifique :** Quel effet de C sur les probabilités et la frontière ?

**Présentation :** La régression logistique est un algorithme supervisé de classification qui transforme une combinaison linéaire des variables en probabilité, P(y=1|x) = 1 / (1 + e^−(β₀ + β·x)) (P = probabilité de la classe 1, x = variables, β₀ et β = coefficients appris par maximum de vraisemblance), avec C (inverse de la régularisation) comme hyperparamètre.

```python
schema('8. Régression logistique', 'coefficients β = 0, probabilité 0.5', 'ajuster β pour maximiser la vraisemblance', 'probabilité puis classe (seuil 0.5)', 'la sigmoïde et la frontière linéaire')
```

**Schéma original :** Le schéma montre le passage de « coefficients β = 0, probabilité 0.5 » (initialisation) à « ajuster β pour maximiser la vraisemblance » (apprentissage) puis à « probabilité puis classe (seuil 0.5) » (sortie), ce qui illustre la sigmoïde et la frontière linéaire.

```python
for C in [0.01, 0.5, 100]:
    m = make_pipeline(StandardScaler(), LogisticRegression(C=C, max_iter=5000, random_state=GRAINE))
    m.fit(Xc_ent, yc_ent)
    proba = m.predict_proba(Xc_ent)[:, 1]
    print('C =', C, '| F1 CV :', cv_f1(m), '| norme des coefficients :', round(np.linalg.norm(m[-1].coef_), 2), '| probabilités extrêmes (<0.05 ou >0.95) :', round(np.mean((proba < .05) | (proba > .95)), 2))
```

**Test de sensibilité et interprétation :** Quand C passe de 0.01 à 0.5 puis 100 la norme des coefficients monte de 0.81 à 2.87 puis 27.82 et la part de probabilités extrêmes de 0.30 à 0.82 puis 0.95 pendant que le F1 CV vaut 0.963, 0.982 puis 0.972, donc un C grand rend la frontière plus tranchée et les probabilités plus sûres d'elles, et un seul découpage ne prouve pas que C=0.5 reste le meilleur ailleurs.

**Présentation :** UMAP est un algorithme non supervisé de réduction de dimension qui construit un graphe de voisins puis place les points en 2D pour conserver ce graphe en minimisant une entropie croisée, Σ[wᵢⱼ·log(wᵢⱼ/vᵢⱼ) + (1 − wᵢⱼ)·log((1 − wᵢⱼ)/(1 − vᵢⱼ))] (wᵢⱼ = poids du lien i–j dans l'espace d'origine, vᵢⱼ = poids dans le plan), les coordonnées étant apprises et n_neighbors et min_dist étant des hyperparamètres.

**Schéma original :** Le schéma montre le passage de « graphe de voisins construit (n_neighbors) » (initialisation) à « placer les points en 2D pour garder le graphe » (apprentissage) puis à « carte 2D compacte » (sortie), ce qui illustre n_neighbors (structure locale ou globale).

**Contextes et choix :** Il convient pour visualiser de grandes bases de clients (marketing) ou de transactions (finance) en gardant un peu de structure globale, face à t-SNE, et il est mal adapté quand on a besoin d'une projection linéaire interprétable, la trustworthiness étant la métrique.

```python
from sklearn.manifold import trustworthiness
fig, axes = plt.subplots(1, 2, figsize=(9, 3.5))
for ax, nv in zip(axes, [8, 30]):
    proj = umap.UMAP(n_components=2, n_neighbors=nv, min_dist=.25, random_state=GRAINE).fit_transform(X_clu)
    print('n_neighbors =', nv, '| trustworthiness (10 voisins) :', round(trustworthiness(X_clu, proj, n_neighbors=10), 3))
    ax.scatter(proj[:, 0], proj[:, 1], s=14); ax.set_title(f'UMAP n_neighbors={nv}')
plt.show()
```

**Test de sensibilité et interprétation :** Avec n_neighbors = 8 l'UMAP met l'accent sur la structure locale (petits amas) alors qu'avec 30 il montre une structure plus globale, ce que la trustworthiness affichée par la cellule ci-dessus permet de vérifier, et une seule graine ne permet pas de conclure sur la taille ou la distance entre amas.

## 25. Bandits manchots

**Variante C — question spécifique :** Quel bras obtient la meilleure moyenne après 800 tirages ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Bandit pédagogique simulé : estimation des probabilités de gain
rng = np.random.default_rng(GRAINE)
probabilites_simulees = np.array([.35, .55, .65, .45])
recompenses = np.zeros(4); nombre_tirages = np.zeros(4)
for etape in range(800):
    bras = etape % 4
    gain = float(rng.random() < probabilites_simulees[bras])
    recompenses[bras] += gain; nombre_tirages[bras] += 1
print('Récompenses moyennes estimées :', np.round(recompenses / nombre_tirages, 3))
```

**Présentation :** Le bandit manchot est un problème de renforcement où un agent tire un bras parmi k et reçoit un gain aléatoire, en estimant la valeur de chaque bras par Q(a) = (somme des gains du bras a)/(nombre de tirages de a) (a = bras), les Q étant appris et le nombre de tirages (le budget) étant le paramètre de conception que l'on règle.

```python
schema('25. Bandits manchots', 'Q = 0 pour chaque bras', 'tirer un bras, observer le gain, mettre Q à jour', 'bras à meilleure moyenne estimée', "l'estimation d'une moyenne par échantillonnage")
```

**Schéma original :** Le schéma montre le passage de « Q = 0 pour chaque bras » (initialisation) à « tirer un bras, observer le gain, mettre Q à jour » (apprentissage) puis à « bras à meilleure moyenne estimée » (sortie), ce qui illustre l'estimation d'une moyenne par échantillonnage.

**Contextes et choix :** Il convient pour choisir entre quatre versions d'une publicité (marketing) ou entre transporteurs (logistique) quand on observe les gains au fil du temps, face à un test A/B classique, et il est mal adapté quand la situation change à chaque décision (il faut alors des états), la récompense moyenne étant la métrique, sur une simulation et non sur des données réelles.

```python
def bandit_tour_de_role(n_tirages, graine, probas=(.35, .55, .65, .45)):
    rng = np.random.default_rng(graine); p = np.array(probas)
    gains = np.zeros(4); nb = np.zeros(4)
    for etape in range(n_tirages):
        bras = etape % 4; gains[bras] += float(rng.random() < p[bras]); nb[bras] += 1
    return gains / nb
for n in [80, 800, 8000]:
    meilleurs = [int(np.argmax(bandit_tour_de_role(n, g))) for g in range(30)]
    print('tirages =', n, '| fréquence où le bras 2 (réellement le meilleur) est identifié, sur 30 graines :', round(np.mean(np.array(meilleurs) == 2), 2))
print('Estimations avec graine 42 et 800 tirages :', np.round(bandit_tour_de_role(800, GRAINE), 3), '-> meilleur bras estimé :', int(np.argmax(bandit_tour_de_role(800, GRAINE))))
```

**Test de sensibilité et interprétation :** Avec 800 tirages et la graine 42 le meilleur bras estimé est le bras 2 (moyennes 0.37, 0.53, 0.63, 0.46) et pour 80, 800 puis 8000 tirages ce bras est identifié dans 80 %, 100 % puis 100 % des 30 graines, donc il faut assez de tirages, mais c'est une simulation avec des probabilités fixées dans le code et non des données réelles.

## 26. Epsilon-greedy

**Variante C — question spécifique :** Quel effet de epsilon sur la concentration des choix ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Exploration epsilon-greedy sur le même bandit simulé
rng = np.random.default_rng(GRAINE)
probabilites_simulees = np.array([.35, .55, .65, .45])
recompenses = np.zeros(4); nombre_tirages = np.zeros(4)
epsilon = .2
for etape in range(1500):
    bras = int(rng.integers(4)) if rng.random() < epsilon or not np.all(nombre_tirages) else int(np.argmax(recompenses / np.maximum(nombre_tirages, 1)))
    gain = float(rng.random() < probabilites_simulees[bras])
    recompenses[bras] += gain; nombre_tirages[bras] += 1
print('Fréquences de sélection :', nombre_tirages.astype(int))
print('Valeurs estimées :', np.round(recompenses / np.maximum(nombre_tirages, 1), 3))
```

**Présentation :** Epsilon-greedy est un algorithme de renforcement qui choisit un bras au hasard avec la probabilité ε (exploration) et sinon le bras de meilleure valeur estimée (exploitation), a = argmax Q(a) avec la probabilité 1 − ε (Q(a) = gain moyen estimé du bras a), les Q étant appris et ε étant l'hyperparamètre.

```python
schema('26. Epsilon-greedy', 'Q = 0, ε fixé', 'explorer avec probabilité ε, sinon exploiter le meilleur Q', 'tirages concentrés sur le meilleur bras', 'le compromis exploration/exploitation (ε)')
```

**Schéma original :** Le schéma montre le passage de « Q = 0, ε fixé » (initialisation) à « explorer avec probabilité ε, sinon exploiter le meilleur Q » (apprentissage) puis à « tirages concentrés sur le meilleur bras » (sortie), ce qui illustre le compromis exploration/exploitation (ε).

**Contextes et choix :** Il convient pour répartir des offres promotionnelles entre quatre messages (marketing) ou des commandes entre transporteurs (logistique) quand on apprend en agissant, face à UCB, et il est mal adapté quand une mauvaise décision coûte très cher, le gain total étant la métrique, sur une simulation.

```python
def eps_greedy(epsilon, n_tirages=1500, graine=GRAINE, probas=(.35, .55, .65, .45)):
    rng = np.random.default_rng(graine); p = np.array(probas)
    gains = np.zeros(4); nb = np.zeros(4)
    for etape in range(n_tirages):
        bras = int(rng.integers(4)) if rng.random() < epsilon or not np.all(nb) else int(np.argmax(gains / np.maximum(nb, 1)))
        gains[bras] += float(rng.random() < p[bras]); nb[bras] += 1
    return nb, gains.sum()
for eps in [0.0, 0.02, 0.2, 0.5]:
    res = [eps_greedy(eps, graine=g) for g in range(20)]
    part = np.mean([r[0][2] / r[0].sum() for r in res]); gain = np.mean([r[1] for r in res])
    print('epsilon =', eps, '| part des tirages sur le meilleur bras :', round(part, 2), '| gain total moyen (20 graines) :', round(gain, 1))
```

**Test de sensibilité et interprétation :** Quand epsilon vaut 0, 0.02, 0.2 puis 0.5 la part des tirages sur le meilleur bras est 0.60, 0.70, 0.79 puis 0.58 et le gain total moyen sur 20 graines 865.4, 910.0, 914.5 puis 856.8, donc un epsilon trop petit reste coincé sur un mauvais bras et un epsilon trop grand gaspille des tirages, et ce résultat vaut pour cette simulation seulement.

## 27. Q-learning

**Variante C — question spécifique :** Étudiez les récompenses des épisodes, pas seulement Q moyen.

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Q-learning manuel — environnement CliffWalking (Gymnasium)
import gymnasium as gym
env = gym.make('CliffWalking-v0')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    for pas in range(120):
        action = int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
        suivant, recompense, termine, tronque, _ = env.step(action)
        cible = recompense + (0 if termine else .95 * np.max(q_table[suivant]))
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat = suivant
        if termine or tronque: break
print('Table Q :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

**Présentation :** Le Q-learning est un algorithme de renforcement hors politique qui apprend la valeur de chaque couple (état, action) par Q(s,a) ← Q(s,a) + α·[r + γ·max Q(s',a') − Q(s,a)] (s = état, a = action, r = récompense, s' = état suivant, α = taux d'apprentissage, γ = facteur d'actualisation), la table Q étant apprise et α, γ et ε étant les hyperparamètres.

```python
schema('27. Q-learning', 'table Q remplie de 0', 'agir (ε-greedy) puis mettre à jour Q avec le meilleur Q suivant', 'politique gloutonne (chemin le plus court)', 'la cible hors politique (max Q)')
```

**Schéma original :** Le schéma montre le passage de « table Q remplie de 0 » (initialisation) à « agir (ε-greedy) puis mettre à jour Q avec le meilleur Q suivant » (apprentissage) puis à « politique gloutonne (chemin le plus court) » (sortie), ce qui illustre la cible hors politique (max Q).

**Contextes et choix :** Il convient pour apprendre une politique de réapprovisionnement ou de routage (logistique) ou d'enchères publicitaires (marketing) avec peu d'états et d'actions, face à SARSA, et il est mal adapté quand une exploration risquée coûte cher (falaise) ou quand il y a trop d'états, la récompense par épisode étant la métrique, sur simulation.

```python
import gymnasium as gym
def entrainer_q_learning(alpha=.15, gamma=.95, eps_min=.08, graine=GRAINE, n_episodes=1100):
    env = gym.make('CliffWalking-v0'); rng = np.random.default_rng(graine)
    q = np.zeros((env.observation_space.n, env.action_space.n)); recompenses_ep = []
    for episode in range(n_episodes):
        etat, _ = env.reset(seed=graine + episode); epsilon = max(eps_min, .8 * (1 - episode / 900)); total = 0
        for pas in range(120):
            action = int(rng.integers(4)) if rng.random() < epsilon else int(np.argmax(q[etat]))
            suivant, r, termine, tronque, _ = env.step(action); total += r
            q[etat, action] += alpha * (r + (0 if termine else gamma * np.max(q[suivant])) - q[etat, action])
            etat = suivant
            if termine or tronque: break
        recompenses_ep.append(total)
    env.close(); return q, np.array(recompenses_ep)
for alpha in [.05, .15, .5]:
    q, rec = entrainer_q_learning(alpha=alpha)
    print('alpha =', alpha, '| récompense moyenne des 100 derniers épisodes :', round(rec[-100:].mean(), 1), '| Q moyen :', round(q.mean(), 2))
```

**Test de sensibilité et interprétation :** La cellule ci-dessus affiche la récompense moyenne des 100 derniers épisodes pour alpha = 0.05, 0.15 puis 0.5 et pas seulement le Q moyen : le Q-learning apprend le chemin optimal au bord de la falaise mais, comme il explore encore, il y tombe parfois et sa récompense d'entraînement reste basse, et ce résultat dépend de la graine et d'une seule exécution.

## 28. SARSA

**Variante C — question spécifique :** SARSA privilégie-t-il une trajectoire plus sûre sous exploration ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# SARSA manuel — même environnement, politique epsilon-greedy suivie
import gymnasium as gym
env = gym.make('CliffWalking-v0')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
def choisir_action(etat, epsilon):
    return int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    action = choisir_action(etat, epsilon)
    for pas in range(120):
        suivant, recompense, termine, tronque, _ = env.step(action)
        action_suivante = choisir_action(suivant, epsilon)
        cible = recompense + (0 if termine else .95 * q_table[suivant, action_suivante])
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat, action = suivant, action_suivante
        if termine or tronque: break
print('Table Q SARSA :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

**Présentation :** SARSA est un algorithme de renforcement sur politique qui met Q à jour avec l'action réellement choisie ensuite, Q(s,a) ← Q(s,a) + α·[r + γ·Q(s',a') − Q(s,a)] (s = état, a = action, r = récompense, s' et a' = état et action suivants, α = taux d'apprentissage, γ = facteur d'actualisation), la table Q étant apprise et α, γ et ε étant les hyperparamètres.

```python
schema('28. SARSA', 'table Q remplie de 0', "agir, choisir l'action suivante puis mettre à jour avec Q(s',a')", 'politique plus prudente (loin de la falaise)', 'la cible sur politique (action réellement choisie)')
```

**Schéma original :** Le schéma montre le passage de « table Q remplie de 0 » (initialisation) à « agir, choisir l'action suivante puis mettre à jour avec Q(s',a') » (apprentissage) puis à « politique plus prudente (loin de la falaise) » (sortie), ce qui illustre la cible sur politique (action réellement choisie).

**Contextes et choix :** Il convient pour piloter un chariot d'entrepôt ou un véhicule de livraison (logistique) quand une erreur d'exploration coûte cher, ou pour ajuster des prix prudemment (commerce), face au Q-learning, et il est mal adapté quand on veut absolument la politique optimale, la récompense par épisode étant la métrique, sur simulation.

```python
def entrainer_sarsa(alpha=.15, gamma=.95, eps_min=.08, graine=GRAINE, n_episodes=1100):
    env = gym.make('CliffWalking-v0'); rng = np.random.default_rng(graine)
    q = np.zeros((env.observation_space.n, env.action_space.n)); recompenses_ep = []
    def choisir(etat, eps): return int(rng.integers(4)) if rng.random() < eps else int(np.argmax(q[etat]))
    for episode in range(n_episodes):
        etat, _ = env.reset(seed=graine + episode); epsilon = max(eps_min, .8 * (1 - episode / 900)); action = choisir(etat, epsilon); total = 0
        for pas in range(120):
            suivant, r, termine, tronque, _ = env.step(action); total += r
            action_suivante = choisir(suivant, epsilon)
            q[etat, action] += alpha * (r + (0 if termine else gamma * q[suivant, action_suivante]) - q[etat, action])
            etat, action = suivant, action_suivante
            if termine or tronque: break
        recompenses_ep.append(total)
    env.close(); return q, np.array(recompenses_ep)
q_ql, rec_ql = entrainer_q_learning(); q_sa, rec_sa = entrainer_sarsa()
print('Récompense moyenne d\'entraînement (100 derniers épisodes) — Q-learning :', round(rec_ql[-100:].mean(), 1), '| SARSA :', round(rec_sa[-100:].mean(), 1))
for eps in [.3]:
    _, r1 = entrainer_q_learning(eps_min=eps); _, r2 = entrainer_sarsa(eps_min=eps)
    print('epsilon final =', eps, '| Q-learning :', round(r1[-100:].mean(), 1), '| SARSA :', round(r2[-100:].mean(), 1))
```

**Test de sensibilité et interprétation :** Sous exploration SARSA est censé privilégier une trajectoire plus sûre, éloignée de la falaise, car il intègre le risque de l'action exploratoire dans Q, ce que la cellule teste en comparant les récompenses d'entraînement des 100 derniers épisodes avec celles du Q-learning pour un epsilon final de 0.08 puis 0.3, mais l'effet dépend de la graine et doit être confirmé sur plusieurs graines (partie finale), donc une seule exécution ne permet pas de conclure.

## 29. DQN — cible conceptuelle

**Variante C — question spécifique :** Pourquoi la cible est-elle égale à la récompense en état terminal ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Cibles DQN : démonstration du calcul de Bellman, SANS réseau neuronal
recompenses = np.array([-1., -100., 0.])
termine = np.array([False, True, False])
meilleure_valeur_suivante = np.array([-3., -2., 5.])
gamma = .9
cibles = recompenses + gamma * (~termine) * meilleure_valeur_suivante
print('Cibles de Bellman calculées :', np.round(cibles, 2))
print('Ce calcul ne constitue pas un entraînement DQN.')
```

**Présentation :** Le DQN est un algorithme de renforcement qui remplace la table Q par un réseau de neurones entraîné vers la cible y = r + γ·(1 − terminé)·max Q_cible(s',a') (r = récompense, γ = facteur d'actualisation, terminé = 1 si l'état est final, Q_cible = réseau cible figé, s' = état suivant), les poids du réseau étant appris et γ, la taille du tampon de rejeu et le taux d'apprentissage étant des hyperparamètres, mais ici seul le calcul de la cible est illustré, sans entraînement.

```python
schema('29. DQN — cible conceptuelle', 'réseau Q avec poids aléatoires', 'calculer la cible de Bellman puis réduire (Q − y)²', 'valeurs Q approchées pour chaque action', 'la cible de Bellman (réseau cible)')
```

**Schéma original :** Le schéma montre le passage de « réseau Q avec poids aléatoires » (initialisation) à « calculer la cible de Bellman puis réduire (Q − y)² » (apprentissage) puis à « valeurs Q approchées pour chaque action » (sortie), ce qui illustre la cible de Bellman (réseau cible).

**Contextes et choix :** Il convient pour des problèmes à très nombreux états comme la gestion dynamique de stock (logistique) ou le trading algorithmique (finance), face au Q-learning tabulaire, et il est mal adapté aux petites données car il demande beaucoup d'expérience simulée, la récompense cumulée étant la métrique.

```python
for gamma in [0.5, 0.9, 0.99]:
    cibles = recompenses + gamma * (~termine) * meilleure_valeur_suivante
    print('gamma =', gamma, '| cibles de Bellman :', np.round(cibles, 2))
```

**Test de sensibilité et interprétation :** Dans un état terminal il n'y a pas de suite, donc la cible est seulement la récompense (−100 dans l'exemple) car le terme γ·max Q(s') est multiplié par 0, et quand gamma passe de 0.5 à 0.9 puis 0.99 la cible du premier état passe de −2.5 à −3.7 puis −3.97 et celle du troisième de 2.5 à 4.5 puis 4.95, mais ce calcul ne constitue pas un DQN entraîné.

```python
# Descente de gradient manuelle sur Diabetes (une seule caractéristique)
from sklearn.linear_model import LinearRegression
x = Xr_ent[:, 2].copy()  # Indice de masse corporelle, caractéristique déjà centrée dans le jeu
x = (x - x.mean()) / x.std()
y = yr_ent.copy()
poids, biais, taux = 0., 0., .05
for iteration in range(1800):
    erreur = (poids * x + biais) - y
    poids -= taux * 2 * np.mean(erreur * x)
    biais -= taux * 2 * np.mean(erreur)
reference = LinearRegression().fit(x.reshape(-1, 1), y)
print('Manuel :', round(poids, 3), round(biais, 3))
print('scikit-learn :', round(reference.coef_[0], 3), round(reference.intercept_, 3))
```

**Annexe 1 — interprétation :** la descente de gradient manuelle retrouve exactement les mêmes valeurs que scikit-learn (pente 45.974 et ordonnée 154.344), ce qui valide l'implémentation, avec un taux d'apprentissage de 0.05 et 1800 itérations.

```python
# K-means manuel sur projection ACP du jeu Wine (sans étiquettes)
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
valeurs = PCA(n_components=2, random_state=GRAINE).fit_transform(X_clu)
rng = np.random.default_rng(GRAINE)
centroides = valeurs[rng.choice(len(valeurs), size=3, replace=False)].copy()
for iteration in range(100):
    distances = ((valeurs[:, None, :] - centroides[None, :, :]) ** 2).sum(axis=2)
    groupes = distances.argmin(axis=1)
    nouveaux = np.array([valeurs[groupes == k].mean(axis=0) if np.any(groupes == k) else centroides[k] for k in range(3)])
    if np.allclose(nouveaux, centroides):
        break
    centroides = nouveaux
print('Itérations manuelles :', iteration + 1)
print('Inertie manuelle :', round(((valeurs - centroides[groupes]) ** 2).sum(), 2))
modele_reference = KMeans(n_clusters=3, n_init=10, random_state=GRAINE).fit(valeurs)
print('Inertie scikit-learn :', round(modele_reference.inertia_, 2))
# L'inertie peut différer si l'initialisation est différente.
```

**Annexe 2 — interprétation :** le K-means manuel converge en 3 itérations avec une inertie de 260.02 contre 259.51 pour scikit-learn, l'écart venant de l'initialisation aléatoire différente (scikit-learn répète l'initialisation 10 fois et garde la meilleure).

### C7 — Bandits simulés
**Réponse :** Epsilon-greedy (0.1) obtient le meilleur gain total moyen (934.2) devant UCB1 (889.3) et le tour de rôle de référence (745.0) sur 20 graines, avec un écart-type plus grand pour epsilon-greedy (38.5), et ce sont des résultats de simulation à ne pas présenter comme des données réelles.

```python
# C7 — Bandits (simulation, 20 graines) : tour de rôle (référence) contre epsilon-greedy et UCB1
def jouer(strategie, graine, n_tirages=1500, probas=(.35, .55, .65, .45)):
    rng = np.random.default_rng(graine); p = np.array(probas); gains = np.zeros(4); nb = np.zeros(4)
    for t in range(n_tirages):
        if strategie == 'tour de rôle': bras = t % 4
        elif strategie == 'epsilon-greedy 0.1': bras = int(rng.integers(4)) if rng.random() < .1 or not np.all(nb) else int(np.argmax(gains / nb))
        else: bras = t if t < 4 else int(np.argmax(gains / nb + np.sqrt(2 * np.log(t + 1) / nb)))
        gains[bras] += float(rng.random() < p[bras]); nb[bras] += 1
    return gains.sum()
resultats_bandit = {s: [jouer(s, g) for g in range(20)] for s in ['tour de rôle', 'epsilon-greedy 0.1', 'UCB1']}
for s, v in resultats_bandit.items(): print(s, '| gain total moyen :', round(np.mean(v), 1), '± écart-type', round(np.std(v), 1))
plt.boxplot(list(resultats_bandit.values()), tick_labels=list(resultats_bandit.keys())); plt.ylabel('Gain total sur 1500 tirages'); plt.title('Bandit simulé — 20 graines'); plt.show()
```

### C7/C8 et (d) — CliffWalking-v0 : agents et comparaison statistique
**Réponse :** La cellule compare sur 10 graines la politique aléatoire, le Q-learning et SARSA (récompense d'entraînement et récompense gloutonne) avec un test de Wilcoxon, et ses sorties servent à vérifier que SARSA est plus sûr pendant l'exploration alors que le Q-learning vise le chemin le plus court, sachant que 10 graines donnent un test peu puissant et que c'est une simulation.

```python
# C7/C8/(d) — CliffWalking-v0 (simulation, 10 graines) : politique aléatoire (référence), Q-learning et SARSA
import gymnasium as gym
from scipy.stats import wilcoxon
def eval_gloutonne(q):
    env = gym.make('CliffWalking-v0'); etat, _ = env.reset(seed=GRAINE); total = 0
    for _ in range(100):
        etat, r, termine, tronque, _ = env.step(int(np.argmax(q[etat]))); total += r
        if termine or tronque: break
    env.close(); return total
def recompense_aleatoire(graine, n_episodes=100):
    env = gym.make('CliffWalking-v0'); rng = np.random.default_rng(graine); totaux = []
    for ep in range(n_episodes):
        etat, _ = env.reset(seed=graine + ep); total = 0
        for _ in range(120):
            etat, r, termine, tronque, _ = env.step(int(rng.integers(4))); total += r
            if termine or tronque: break
        totaux.append(total)
    env.close(); return np.mean(totaux)
res = {'Aléatoire (référence)': {'entrainement': [], 'glouton': []}, 'Q-learning': {'entrainement': [], 'glouton': []}, 'SARSA': {'entrainement': [], 'glouton': []}}
courbes = {'Q-learning': [], 'SARSA': []}
for g in range(10):
    res['Aléatoire (référence)']['entrainement'].append(recompense_aleatoire(g))
    for nom, fonction in [('Q-learning', entrainer_q_learning), ('SARSA', entrainer_sarsa)]:
        q, rec = fonction(graine=g)
        res[nom]['entrainement'].append(rec[-100:].mean()); res[nom]['glouton'].append(eval_gloutonne(q)); courbes[nom].append(rec)
for nom, d in res.items():
    gl = ('| récompense gloutonne (sans exploration) : %.1f ± %.1f' % (np.mean(d['glouton']), np.std(d['glouton']))) if d['glouton'] else ''
    print(nom, '| récompense d\'entraînement (100 derniers épisodes) : %.1f ± %.1f' % (np.mean(d['entrainement']), np.std(d['entrainement'])), gl)
for cle in ['entrainement', 'glouton']:
    a, b = np.array(res['Q-learning'][cle]), np.array(res['SARSA'][cle])
    try: print('Test de Wilcoxon Q-learning contre SARSA (', cle, ') : p =', round(wilcoxon(a, b).pvalue, 4))
    except ValueError: print('Test de Wilcoxon (', cle, ') impossible : toutes les différences sont nulles.')
for nom, c in courbes.items(): plt.plot(np.convolve(np.mean(c, axis=0), np.ones(50) / 50, mode='valid'), label=nom)
plt.xlabel('Épisode'); plt.ylabel('Récompense moyenne (10 graines, lissée)'); plt.legend(); plt.title('CliffWalking-v0 — apprentissage'); plt.show()
```
